```python
# 8 - Faça um programa que retorne True se duas strings lidas do teclado são palíndromas mútuas, ou seja, se uma é igual à outra quando lida de traz para frente ou False, se não forem.
# Por exemplo: amor e roma
```

```python
p1= "Amor".lower()
p2= "Roza".lower()
```

```python
if p1[::-1]== p2:
  print("True")
elif p2[::-1]==p1:
    print("True")
else:
  print("False")
```
