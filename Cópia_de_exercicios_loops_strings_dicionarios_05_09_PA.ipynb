### 1. **Contagem de vogais**: Peça uma palavra e conte quantas vezes cada vogal aparece (usar `for` ou `while`).

```python
frase = (input("Digite a frase: ")).lower()
vogais = {'a':0, 'e':0, 'i':0, 'o':0, 'u':0}

for letra in frase:
    if letra in vogais:
      vogais[letra] += 1
print("a quantidade de vogais é ", vogais)

```

### 2. **Inversão de string**: Solicite uma frase e mostre-a invertida **sem usar slicing** (`[::-1]` não vale). Use `while`.

```python
frase = str(input("digite uma frase: "))
invertido = ""
i = len(frase) - 1
while i >= 0:
  invertido +=frase[i]
  i -=1
print("a frase invertida é: ", invertido)
```

### 4. **Palavras únicas**: Peça uma frase e mostre as palavras que aparecem **apenas uma vez**.

```python
frase= input("digite uma frase: ").lower()
frequencia = {}
chocolate = frase.split()
for chocolat in chocolate:
   frequencia[chocolat] = frequencia.get(chocolat, 0) + 1
print("palavras unicas:")
for chocolat, contagem in frequencia.items():
    if contagem == 1:
        print(chocolat)
```

### 6. **Palíndromo**: Verifique se uma palavra é um **palíndromo** sem usar funções prontas de reversão.

```python
pala=input("Digite uma palavra: ").upper()
for i in range(len(pala)//2):
   if pala[i] != pala[len(pala)-1-i]:
    print(f"'{pala}'""Não é palindroma!")
    break
else:
    print(f"'{pala}'"" É palindroma! ")
```

### 7. **Enigma da senha**: Peça uma string e crie uma nova palavra pegando apenas as letras nas **posições ímpares**.

### 8. **Expansão de abreviações**: Dado um dicionário de abreviações, substitua manualmente cada ocorrência em uma frase.

(Ex: (vc: você, blz: beleza ...

```python
abreviacoes={'vc':'voce',
             'blz':'beleza',
             'foto':'fotografia'}

primeirafrase = " Ok, blz vc é muito boa em tirar foto."
fraseminu= primeirafrase.lower()
sepapalavras = fraseminu.split(' ')
novafrase = []
for i in sepapalavras:
  pal= abreviacoes.get(i,i)
  novafrase.append(pal)
frasefim = ' '.join(novafrase)
print(f"A frase nova é: {frasefim}")

```

### 9. **Criptografia simples**: Implemente uma cifra de substituição onde cada letra é trocada pela **próxima no alfabeto** (z → a).

```python
alfabeto="abcdefghijklmnopqrstuvwxyz"
palavra="ela trabalha bem"
novapalavra = ""
for i in palavra:
  if i.isalpha():
    maius = i.isupper()
    minus = i.lower()
    posicao = alfabeto.find(minus)
    if minus == "z":
        novaletradepoisoriginal = "a"
    else:
       novaletradepoisoriginal = alfabeto[posicao + 1]
    if maius:
      novaletradepoisoriginal= novaletradepoisoriginal.upper()
    novapalavra += novaletradepoisoriginal
  else:
    novapalavra += i
print(novapalavra)
```

### 10. **Anagrama**: Verifique se duas palavras são **anagramas**, usando dicionário para contar ocorrências.

Anagrama é uma palavra ou frase formada pela reorganização das letras de outra palavra ou frase, usando exatamente as mesmas letras, cada uma uma única vez.

```python
pala1="amor"
pala2="roma"
pala1=pala1.lower()
pala2=pala2.lower()
if len(pala1) != len(pala2):
  print("A palavra 1 e 2 não são anagramas")
else:
  freq1={}
  freq2={}
  for i in pala1:
    freq1[i]=freq1.get(i,0)+1
  for i in pala2:
    freq2[i]=freq2.get(i,0)+1
  if freq1==freq2:
    print(f"As palavras {pala1}, {pala2} são anagramas")
  else:
    print(f"As palavras {pala1}, {pala2} não são anagramas")
```
